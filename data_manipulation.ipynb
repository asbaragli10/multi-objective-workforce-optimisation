```python
import pandas as pd
```

```python
data= pd.read_csv(r"C:\Users\andre\OneDrive\Documenti\Heuristics\MT_SA_Jigs\Data\Berwick_pred.csv")
data.head()
```

```python
deadline_map = {
    'Z1-01': '16-May-25',
    'Z1-00': '12-May-25',
    'Z2-01': '19-May-25',
    'Z2-00': '19-May-25',
    'Z4-01': '19-May-25',
    'Z4-00': '18-Jun-25',
    'Z3-01': '10-Jun-25',  
    'Z3-00': '02-Jul-25'  
}



# Extract the prefix (first two parts)
data['prefix'] = data['AssemblyRef'].str.extract(r'^(Z\d-\d{2})')

# Map the deadline
data['deadline'] = data['prefix'].map(deadline_map).fillna('1-Jun-25')

# Drop prefix column if not needed
data = data.drop(columns='prefix')
```

```python
selected_df = data[['AssemblyRef', 'Len', 'Coil', 'Work Time', 'DIMPLE_count', 'deadline']].copy()
selected_df = selected_df.rename(columns={'Work Time': 'Howick Work Time'})
selected_df.head()


```

```python
selected_df.to_csv("Berwick_full_deadlines.csv")
# This output can be used for the howick and the jigs 
```
